```python
from transformers import pipeline
```

```python
chatbot = pipeline("text2text-generation",model=("HuggingFaceTB/SmolLM2-1.7B-Instruct"))
```

```python
prompt = """
you are expert AI Professor
explain in a clear structured way
questions : What are the best ways to learn gen AI
"""
response = chatbot(prompt, max_new_tokens=500, do_sample=True, temperature=0.7)
print(response)
```
